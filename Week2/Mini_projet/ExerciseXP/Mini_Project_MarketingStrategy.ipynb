```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")


file_path = "/content/US Superstore data.xls"

df = pd.read_excel(file_path)

print("Aperçu du dataset :")
print(df.head())

print("\nDimensions du dataset :")
print(df.shape)

print("\nColonnes du dataset :")
print(df.columns)

print("\nInformations générales :")
print(df.info())

df.columns = df.columns.str.strip()

print("\nNombre de doublons :", df.duplicated().sum())

df = df.drop_duplicates()

print("\nValeurs manquantes par colonne :")
print(df.isnull().sum())

if "Postal Code" in df.columns:
    df["Postal Code"] = df["Postal Code"].fillna(0)

df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")

df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month-Year"] = df["Order Date"].dt.to_period("M")

df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100

print("\nAperçu après nettoyage :")
print(df.head())


state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False)

print("\nTop 10 des États avec le plus de ventes :")
print(state_sales.head(10))

plt.figure(figsize=(12, 6))
state_sales.head(10).plot(kind="bar")
plt.title("Top 10 des États avec le plus de ventes")
plt.xlabel("État")
plt.ylabel("Ventes totales")
plt.xticks(rotation=45)
plt.show()


ny_ca = df[df["State"].isin(["New York", "California"])]

ny_ca_summary = ny_ca.groupby("State")[["Sales", "Profit"]].sum()

print("\nComparaison New York vs California :")
print(ny_ca_summary)

ny_ca_summary.plot(kind="bar", figsize=(8, 5))
plt.title("Comparaison des ventes et bénéfices : New York vs California")
plt.xlabel("État")
plt.ylabel("Montant")
plt.xticks(rotation=0)
plt.show()

ny_clients = df[df["State"] == "New York"]

top_ny_client = ny_clients.groupby("Customer Name")[["Sales", "Profit"]].sum()
top_ny_client = top_ny_client.sort_values(by="Profit", ascending=False)

print("\nClient le plus rentable à New York :")
print(top_ny_client.head(1))

print("\nTop 10 clients à New York selon le profit :")
print(top_ny_client.head(10))

plt.figure(figsize=(12, 6))
top_ny_client["Profit"].head(10).plot(kind="bar")
plt.title("Top 10 des clients les plus rentables à New York")
plt.xlabel("Client")
plt.ylabel("Profit")
plt.xticks(rotation=45, ha="right")
plt.show()


state_profitability = df.groupby("State").agg({
    "Sales": "sum",
    "Profit": "sum"
})

state_profitability["Profit Margin"] = (
    state_profitability["Profit"] / state_profitability["Sales"]
) * 100

state_profitability = state_profitability.sort_values(by="Profit Margin", ascending=False)

print("\nRentabilité par État :")
print(state_profitability.head(10))

plt.figure(figsize=(12, 6))
state_profitability["Profit Margin"].head(10).plot(kind="bar")
plt.title("Top 10 des États avec la meilleure marge bénéficiaire")
plt.xlabel("État")
plt.ylabel("Marge bénéficiaire (%)")
plt.xticks(rotation=45)
plt.show()

print("\nÉtats les moins rentables :")
print(state_profitability.tail(10))


customer_profit = df.groupby("Customer Name")["Profit"].sum().sort_values(ascending=False)

customer_profit_df = customer_profit.reset_index()
customer_profit_df["Cumulative Profit"] = customer_profit_df["Profit"].cumsum()
customer_profit_df["Cumulative Profit Percentage"] = (
    customer_profit_df["Cumulative Profit"] / customer_profit_df["Profit"].sum()
) * 100
customer_profit_df["Customer Percentage"] = (
    (customer_profit_df.index + 1) / len(customer_profit_df)
) * 100

print("\nAnalyse Pareto clients/profits :")
print(customer_profit_df.head())

plt.figure(figsize=(10, 6))
plt.plot(
    customer_profit_df["Customer Percentage"],
    customer_profit_df["Cumulative Profit Percentage"],
    marker="o"
)
plt.axhline(y=80, color="red", linestyle="--", label="80% des profits")
plt.axvline(x=20, color="green", linestyle="--", label="20% des clients")
plt.title("Courbe cumulative des profits par client")
plt.xlabel("Pourcentage cumulé des clients")
plt.ylabel("Pourcentage cumulé des profits")
plt.legend()
plt.grid(True)
plt.show()

top_20_percent_count = int(len(customer_profit_df) * 0.2)
top_20_profit = customer_profit_df.head(top_20_percent_count)["Profit"].sum()
total_profit = customer_profit_df["Profit"].sum()
top_20_profit_percentage = (top_20_profit / total_profit) * 100

print(f"\nLes 20% meilleurs clients génèrent {top_20_profit_percentage:.2f}% des profits.")


city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

print("\nTop 20 villes en chiffre d'affaires :")
print(city_sales.head(20))

plt.figure(figsize=(12, 7))
city_sales.head(20).plot(kind="bar")
plt.title("Top 20 villes en chiffre d'affaires")
plt.xlabel("Ville")
plt.ylabel("Ventes totales")
plt.xticks(rotation=45, ha="right")
plt.show()


city_profit = df.groupby("City")["Profit"].sum().sort_values(ascending=False)

print("\nTop 20 villes en bénéfice :")
print(city_profit.head(20))

plt.figure(figsize=(12, 7))
city_profit.head(20).plot(kind="bar")
plt.title("Top 20 villes en bénéfice")
plt.xlabel("Ville")
plt.ylabel("Bénéfice total")
plt.xticks(rotation=45, ha="right")
plt.show()


city_profitability = df.groupby("City").agg({
    "Sales": "sum",
    "Profit": "sum"
})

city_profitability["Profit Margin"] = (
    city_profitability["Profit"] / city_profitability["Sales"]
) * 100

city_profitability = city_profitability.sort_values(by="Profit Margin", ascending=False)

print("\nTop 20 villes les plus rentables :")
print(city_profitability.head(20))

plt.figure(figsize=(12, 7))
city_profitability["Profit Margin"].head(20).plot(kind="bar")
plt.title("Top 20 villes avec la meilleure marge bénéficiaire")
plt.xlabel("Ville")
plt.ylabel("Marge bénéficiaire (%)")
plt.xticks(rotation=45, ha="right")
plt.show()


top_customers_sales = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False)

print("\nTop 20 clients en termes de ventes :")
print(top_customers_sales.head(20))

plt.figure(figsize=(12, 7))
top_customers_sales.head(20).plot(kind="bar")
plt.title("Top 20 clients en termes de ventes")
plt.xlabel("Client")
plt.ylabel("Ventes totales")
plt.xticks(rotation=45, ha="right")
plt.show()


customer_sales = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False)

customer_sales_df = customer_sales.reset_index()
customer_sales_df["Cumulative Sales"] = customer_sales_df["Sales"].cumsum()
customer_sales_df["Cumulative Sales Percentage"] = (
    customer_sales_df["Cumulative Sales"] / customer_sales_df["Sales"].sum()
) * 100
customer_sales_df["Customer Percentage"] = (
    (customer_sales_df.index + 1) / len(customer_sales_df)
) * 100

plt.figure(figsize=(10, 6))
plt.plot(
    customer_sales_df["Customer Percentage"],
    customer_sales_df["Cumulative Sales Percentage"],
    marker="o"
)
plt.axhline(y=80, color="red", linestyle="--", label="80% des ventes")
plt.axvline(x=20, color="green", linestyle="--", label="20% des clients")
plt.title("Courbe cumulative des ventes par client")
plt.xlabel("Pourcentage cumulé des clients")
plt.ylabel("Pourcentage cumulé des ventes")
plt.legend()
plt.grid(True)
plt.show()

top_20_sales_count = int(len(customer_sales_df) * 0.2)
top_20_sales = customer_sales_df.head(top_20_sales_count)["Sales"].sum()
total_sales = customer_sales_df["Sales"].sum()
top_20_sales_percentage = (top_20_sales / total_sales) * 100

print(f"\nLes 20% meilleurs clients génèrent {top_20_sales_percentage:.2f}% des ventes.")


monthly_trends = df.groupby("Order Month-Year")[["Sales", "Profit"]].sum()

plt.figure(figsize=(12, 6))
plt.plot(monthly_trends.index.astype(str), monthly_trends["Sales"], label="Ventes")
plt.plot(monthly_trends.index.astype(str), monthly_trends["Profit"], label="Bénéfices")
plt.title("Évolution mensuelle des ventes et bénéfices")
plt.xlabel("Mois")
plt.ylabel("Montant")
plt.xticks(rotation=90)
plt.legend()
plt.grid(True)
plt.show()


print("\n RECOMMANDATIONS MARKETING ")

print("\n1. États à privilégier")
print("Les États avec les ventes et profits les plus élevés doivent être prioritaires pour les campagnes marketing.")
print("Il faut notamment observer les performances de California et New York.")

print("\n2. Villes à cibler")
print("Les villes présentes à la fois dans le top des ventes et le top des bénéfices sont les plus intéressantes.")
print("Elles doivent être ciblées avec des campagnes personnalisées.")

print("\n3. Clients à forte valeur")
print("Les meilleurs clients en ventes et en profits doivent recevoir des offres de fidélisation.")
print("Cela peut inclure des remises personnalisées, des programmes VIP ou un suivi commercial dédié.")

print("\n4. Attention aux zones peu rentables")
print("Certaines villes ou certains États peuvent générer beaucoup de ventes mais peu de bénéfices.")
print("Il faut analyser les remises, les coûts ou les catégories vendues dans ces zones.")

print("\n5. Principe de Pareto")
print(f"Les 20% meilleurs clients génèrent {top_20_sales_percentage:.2f}% des ventes.")
print(f"Les 20% meilleurs clients génèrent {top_20_profit_percentage:.2f}% des profits.")
print("Si ces pourcentages sont proches de 80%, le principe de Pareto s'applique fortement.")
print("Sinon, les ventes et profits sont plus répartis entre les clients.")
```
