1. Pourquoi la visualisation des données est importante ?

La visualisation des données est importante parce qu’elle permet de représenter les données sous forme de graphiques, de tableaux ou de diagrammes afin de les rendre plus faciles à comprendre.

Lorsqu’on travaille avec beaucoup de données, il est parfois difficile de voir directement les tendances ou les problèmes dans un tableau. Grâce à la visualisation, on peut rapidement identifier :

- les tendances générales ;
- les différences entre plusieurs groupes ;
- les valeurs anormales ;
- les évolutions dans le temps ;
- les relations entre plusieurs variables.

Par exemple, au lieu de lire une longue colonne de chiffres sur les ventes d’une entreprise, un graphique peut montrer rapidement si les ventes augmentent ou diminuent.

La visualisation aide donc à mieux analyser les données et à prendre de meilleures décisions.

2. Utilité d’un graphique linéaire

Un graphique linéaire est utilisé pour montrer l’évolution d’une donnée dans le temps.

Il est très utile lorsqu’on veut observer une tendance, par exemple :

- l’évolution des ventes par mois ;
- le nombre d’accidents par année ;
- la température pendant une semaine ;
- l’évolution du salaire moyen dans le temps ;
- le nombre de visiteurs d’un site web par jour.

Dans un graphique linéaire, les points sont reliés par une ligne. Cela permet de voir facilement si la valeur augmente, diminue ou reste stable.

```python

import matplotlib.pyplot as plt

jours = ["Lundi", "Mardi", "Mercredi", "Jeudi", "Vendredi", "Samedi", "Dimanche"]

temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(8, 5))
plt.plot(jours, temperatures, marker="o")

plt.title("Variation de la température sur une semaine")
plt.xlabel("Jour")
plt.ylabel("Température (°F)")

plt.grid(True)
plt.show()
```

```python

import matplotlib.pyplot as plt

mois = ["Janvier", "Février", "Mars", "Avril", "Mai"]

ventes = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(8, 5))
plt.bar(mois, ventes)

plt.title("Ventes mensuelles d'un magasin de détail")
plt.xlabel("Mois")
plt.ylabel("Montant des ventes ($)")

plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "/content/Student Mental health.csv"
df = pd.read_csv(file_path)

print("Aperçu du dataset :")
print(df.head())

print("\nColonnes du dataset :")
print(df.columns)

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="What is your CGPA?",
    color="skyblue"
)

plt.title("Distribution des catégories de CGPA")
plt.xlabel("Catégories de CGPA")
plt.ylabel("Nombre d'étudiants")

plt.xticks(rotation=45)
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "/content/Student Mental health.csv"
df = pd.read_csv(file_path)

print("Aperçu du dataset :")
print(df.head())

print("\nColonnes du dataset :")
print(df.columns)

df["Panic_Attack_Numeric"] = df["Do you have Panic attack?"].map({
    "Yes": 1,
    "No": 0
})

print("\nVérification de la conversion :")
print(df[["Age", "Do you have Panic attack?", "Panic_Attack_Numeric"]].head())

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Panic_Attack_Numeric",
    hue="Do you have Panic attack?",
    s=100
)

plt.title("Relation entre l'âge et les crises de panique")
plt.xlabel("Âge")
plt.ylabel("Crise de panique : Non = 0, Oui = 1")

plt.yticks([0, 1], ["Non", "Oui"])
plt.grid(True)

plt.show()
```
