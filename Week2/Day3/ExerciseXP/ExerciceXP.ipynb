Exercice 3

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

moyenne = 50
ecart_type = 10

x = np.linspace(10, 90, 1000)

y = norm.pdf(x, loc=moyenne, scale=ecart_type)

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title("Distribution normale avec moyenne = 50 et écart type = 10")
plt.xlabel("Valeurs")
plt.ylabel("Densité de probabilité")
plt.grid(True)
plt.show()
```

Exercice 4

```python

import numpy as np
from scipy.stats import ttest_ind

np.random.seed(42)

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_statistic, p_value = ttest_ind(data1, data2)

print("Moyenne de data1 :", np.mean(data1))
print("Moyenne de data2 :", np.mean(data2))

print("\nRésultat du test t :")
print("Statistique t :", t_statistic)
print("Valeur p :", p_value)

alpha = 0.05

if p_value < alpha:
    print("\nConclusion : Il existe une différence significative entre les deux groupes.")
else:
    print("\nConclusion : Il n'existe pas de différence significative entre les deux groupes.")
```

Exercice 5

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

result = linregress(house_sizes, house_prices)

slope = result.slope
intercept = result.intercept
r_value = result.rvalue
p_value = result.pvalue
std_error = result.stderr

house_size_90 = 90
predicted_price = slope * house_size_90 + intercept

print("Pente :", slope)
print("Ordonnée à l'origine :", intercept)
print("Prix estimé pour une maison de 90 m² :", predicted_price)

print("\nCoefficient de corrélation :", r_value)
print("Valeur p :", p_value)
print("Erreur standard :", std_error)


plt.figure(figsize=(8, 5))

plt.scatter(house_sizes, house_prices, label="Données réelles")

regression_line = slope * house_sizes + intercept
plt.plot(house_sizes, regression_line, label="Droite de régression")

plt.scatter(house_size_90, predicted_price, label="Prédiction pour 90 m²")

plt.title("Régression linéaire : taille de la maison et prix")
plt.xlabel("Taille de la maison en m²")
plt.ylabel("Prix de la maison")
plt.legend()
plt.grid(True)
plt.show()
```

Exercice 6

```python

from scipy.stats import f_oneway

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print("Valeur F :", f_value)
print("Valeur p :", p_value)

alpha = 0.05

if p_value < alpha:
    print("Conclusion : Les engrais ont des effets significativement différents sur la croissance des plantes.")
else:
    print("Conclusion : Il n'y a pas de différence significative entre les effets des engrais.")
```

Exercice 7

```python


from scipy.stats import binom
import matplotlib.pyplot as plt
import numpy as np

n = 10
p = 0.5
k = 5
probabilite_5_faces = binom.pmf(k, n, p)

print("Probabilité d'obtenir exactement 5 faces en 10 lancers :")
print(probabilite_5_faces)

successes = np.arange(0, n + 1)
probabilities = binom.pmf(successes, n, p)

print("\nProbabilités pour chaque nombre de faces :")
for success, probability in zip(successes, probabilities):
    print(f"{success} face(s) : {probability}")

plt.figure(figsize=(8, 5))
plt.bar(successes, probabilities)

plt.title("Distribution binomiale : nombre de faces en 10 lancers")
plt.xlabel("Nombre de faces")
plt.ylabel("Probabilité")
plt.xticks(successes)
plt.grid(axis="y")
plt.show()
```

```python

import pandas as pd
from scipy.stats import pearsonr, spearmanr

data = pd.DataFrame({
    "age": [23, 25, 30, 35, 40],
    "income": [35000, 40000, 50000, 60000, 70000]
})

print("Dataset :")
print(data)

pearson_corr, pearson_p_value = pearsonr(data["age"], data["income"])

spearman_corr, spearman_p_value = spearmanr(data["age"], data["income"])

print("\nCorrélation de Pearson :")
print("Coefficient :", pearson_corr)
print("Valeur p :", pearson_p_value)

print("\nCorrélation de Spearman :")
print("Coefficient :", spearman_corr)
print("Valeur p :", spearman_p_value)
```
