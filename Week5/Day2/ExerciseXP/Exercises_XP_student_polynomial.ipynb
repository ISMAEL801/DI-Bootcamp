## 🌟 Exercise 3: Creating the Dataset and Visualizing the Data

**As stated in the exercise**  
Objective. Generate a dataset with noise and visualize it.  
Instructions. Import numpy, matplotlib.pyplot, and mean_squared_error from sklearn.metrics. Create 20 points using \(y = -x^2\) with Gaussian noise \(\mathcal{N}(0, 0.05)\). Plot the points with a scatter plot. Split into a training set with the first 12 points and a test set with the last 8 points.


```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

np.random.seed(0)

x = np.arange(-1, 1, 0.1)

y = -x**2 + np.random.normal(0, 0.05, len(x))

print("Valeurs de x :")
print(x)

print("\nValeurs de y :")
print(y)

print("\nNombre de points :", len(x))
```

```python

plt.figure(figsize=(8, 5))

plt.scatter(x, y, label="Données bruitées")

plt.title("Visualisation des données bruitées")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python

x_train = x[:12]
y_train = y[:12]

x_test = x[12:]
y_test = y[12:]

print("Nombre de points d'entraînement :", len(x_train))
print("Nombre de points de test :", len(x_test))

print("\nx_train :")
print(x_train)

print("\nx_test :")
print(x_test)
```

**Learning point**  
Visualizing noisy data helps separate signal from noise. Splitting into train and test allows you to detect when a model overfits noise instead of learning the underlying pattern.


## 🌟 Exercise 4: Fitting Polynomial Models of Different Degrees

**As stated in the exercise**  
Objective: Fit polynomial models of varying degrees and observe overfitting.  

Instructions: Define `polynomial_fit(degree)` that returns the fitted polynomial coefficients from training data. Define `plot_polyfit(degree)` that draws the training set, the test set, and the fitted polynomial curve. Visualize fits for degrees 1, 7, and 11. Observe behavior on training and test sets as degree increases.


```python


def polynomial_fit(degree):
    """
    Cette fonction prend un degré polynomial en entrée
    et retourne un polynôme ajusté sur les données d'entraînement.
    """

    coefficients = np.polyfit(x_train, y_train, degree)

    polynomial = np.poly1d(coefficients)

    return polynomial
```

```python


def plot_polyfit(degree):
    """
    Cette fonction affiche :
    - les données d'entraînement
    - les données de test
    - la courbe polynomiale ajustée
    """

    polynomial = polynomial_fit(degree)

    x_line = np.linspace(-1, 1, 200)
    y_line = polynomial(x_line)

    plt.figure(figsize=(8, 5))

    plt.scatter(x_train, y_train, label="Données d'entraînement", marker="o")
    plt.scatter(x_test, y_test, label="Données de test", marker="x")
    plt.plot(x_line, y_line, label=f"Polynôme degré {degree}")

    plt.title(f"Ajustement polynomial de degré {degree}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)

    plt.show()
```

```python


plot_polyfit(1)
plot_polyfit(7)
plot_polyfit(11)
```

**Learning point**  
Higher degree polynomials can perfectly fit training data yet fail to generalize. This is overfitting. You should always check test error, not just training error.

##Observation

Le polynôme de degré 1 est trop simple pour représenter les données, car les points suivent une forme de parabole. Il fait donc du sous-apprentissage.

Le polynôme de degré 7 suit beaucoup mieux les données d'entraînement, mais il commence à devenir très flexible.

Le polynôme de degré 11 est très complexe. Il peut suivre presque parfaitement les points d'entraînement, mais il risque de mal prédire les points de test. Cela montre un risque de surapprentissage.

## 🌟 Exercise 5: Cross-Validation to Find the Optimal Degree

**As stated in the exercise**  
Objective: Use cross-validation to determine the best polynomial degree.  

Instructions: Loop degrees from 1 to 11. For each degree, compute the RMSE on the training and test sets. Store results in a list. Plot RMSE vs degree with a logarithmic y axis. Identify the degree minimizing test RMSE and confirm it matches the true model \(y=-x^2\).


```python


results = []

for degree in range(1, 12):

    polynomial = polynomial_fit(degree)

    y_train_pred = polynomial(x_train)

    y_test_pred = polynomial(x_test)

    train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))

    results.append([degree, train_rmse, test_rmse])

print("Résultats :")
for result in results:
    print(f"Degré {result[0]} | RMSE Train : {result[1]:.4f} | RMSE Test : {result[2]:.4f}")
```

```python


import pandas as pd

results_df = pd.DataFrame(
    results,
    columns=["Degree", "Train_RMSE", "Test_RMSE"]
)

display(results_df)
```

```python


plt.figure(figsize=(9, 5))

plt.plot(results_df["Degree"], results_df["Train_RMSE"], marker="o", label="RMSE entraînement")
plt.plot(results_df["Degree"], results_df["Test_RMSE"], marker="o", label="RMSE test")

plt.yscale("log")

plt.title("RMSE selon le degré du polynôme")
plt.xlabel("Degré du polynôme")
plt.ylabel("RMSE")
plt.xticks(range(1, 12))
plt.legend()
plt.grid(True)

plt.show()
```

```python


best_row = results_df.loc[results_df["Test_RMSE"].idxmin()]

best_degree = int(best_row["Degree"])
best_test_rmse = best_row["Test_RMSE"]

print("Meilleur degré polynomial :", best_degree)
print("RMSE minimale sur le test :", best_test_rmse)
```

##CONCLUSION


Le meilleur degré polynomial est celui qui obtient la plus petite RMSE sur l'ensemble de test. Comme les données ont été générées à partir de la fonction y = -x², on s'attend à ce qu'un polynôme de degré proche de 2 soit le plus adapté.

Un degré trop faible, comme 1, produit un sous-apprentissage parce qu'il ne peut pas représenter correctement la forme parabolique des données.

Un degré trop élevé, comme 7 ou 11, peut produire un surapprentissage. Dans ce cas, le modèle apprend trop les détails et le bruit des données d'entraînement, mais il généralise moins bien sur les données de test.

Le meilleur modèle est donc celui qui trouve un bon équilibre entre erreur d'entraînement faible et erreur de test faible.
