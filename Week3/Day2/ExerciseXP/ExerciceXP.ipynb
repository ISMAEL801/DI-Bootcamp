```python

import numpy as np

matrix = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

print("Matrice :")
print(matrix)

determinant = np.linalg.det(matrix)

print("\nDéterminant de la matrice :")
print(determinant)

if determinant != 0:
    inverse_matrix = np.linalg.inv(matrix)

    print("\nInverse de la matrice :")
    print(inverse_matrix)
else:
    print("\nLa matrice n'est pas inversible car son déterminant est égal à 0.")
```

```python
import numpy as np

np.random.seed(42)

productivity_before = np.random.normal(loc=50, scale=10, size=30)

productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)

difference = mean_after - mean_before

print("Moyenne avant formation :", mean_before)
print("Moyenne après formation :", mean_after)
print("Différence moyenne :", difference)



if difference > 0:
    print("\nConclusion : La productivité moyenne a augmenté après la formation.")
    print("Cela suggère que le programme de formation peut être efficace.")
else:
    print("\nConclusion : La productivité moyenne n'a pas augmenté.")
    print("Le programme ne semble pas avoir amélioré la productivité.")
```

```python
import pandas as pd
import numpy as np

dates = pd.date_range(start="2023-01-01", end="2023-12-31", freq="D")

np.random.seed(42)
values = np.random.randint(1, 100, len(dates))

time_series = pd.DataFrame({
    "Date": dates,
    "Value": values
})

time_series = time_series.set_index("Date")

print("Aperçu des données de séries temporelles :")
print(time_series.head())

jan_mars = time_series["2023-01-01":"2023-03-31"]

avril_juin = time_series["2023-04-01":"2023-06-30"]

juillet_septembre = time_series["2023-07-01":"2023-09-30"]

octobre_decembre = time_series["2023-10-01":"2023-12-31"]

print("\nJanvier à mars :")
print(jan_mars.head())

print("\nAvril à juin :")
print(avril_juin.head())

print("\nJuillet à septembre :")
print(juillet_septembre.head())

print("\nOctobre à décembre :")
print(octobre_decembre.head())
```
