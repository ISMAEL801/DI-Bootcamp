```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import f_oneway, pearsonr, spearmanr

sns.set(style="whitegrid")



file_path = "/content/train.csv"

df = pd.read_csv(file_path)

print("Aperçu du dataset :")
print(df.head())

print("\nDimensions du dataset :")
print(df.shape)

print("\nColonnes du dataset :")
print(df.columns)

print("\nInformations générales :")
print(df.info())

print("\nValeurs manquantes par colonne :")
print(df.isnull().sum())

print("\nStatistiques descriptives :")
print(df.describe())


target = "price_range"

print("\nVariable cible :", target)
print("\nRépartition des classes de prix :")
print(df[target].value_counts().sort_index())

print("\nTypes de données :")
print(df.dtypes)

print("\nNombre de doublons avant suppression :", df.duplicated().sum())
df = df.drop_duplicates()
print("Nombre de doublons après suppression :", df.duplicated().sum())

for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype in ["int64", "float64"]:
            df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])

print("\nValeurs manquantes après traitement :")
print(df.isnull().sum())

categorical_cols = df.select_dtypes(include=["object"]).columns

print("\nColonnes catégorielles détectées :")
print(categorical_cols)

if len(categorical_cols) > 0:
    df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

print("\nAperçu après prétraitement :")
print(df.head())


numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

print("\nColonnes numériques :")
print(numeric_columns)

stat_results = []

for col in numeric_columns:
    data = df[col].dropna()

    moyenne = np.mean(data)
    mediane = np.median(data)
    mode = stats.mode(data, keepdims=True).mode[0]
    etendue = np.max(data) - np.min(data)
    variance = np.var(data)
    ecart_type = np.std(data)
    skewness = stats.skew(data)
    kurtosis = stats.kurtosis(data)

    stat_results.append({
        "Feature": col,
        "Moyenne": moyenne,
        "Médiane": mediane,
        "Mode": mode,
        "Étendue": etendue,
        "Variance": variance,
        "Écart type": ecart_type,
        "Asymétrie": skewness,
        "Kurtosis": kurtosis
    })

stats_df = pd.DataFrame(stat_results)

print("\nAnalyse statistique détaillée :")
print(stats_df)

print("\nTests ANOVA entre les groupes de price_range :")

anova_results = []

for col in numeric_columns:
    if col != target:
        groups = []

        for price_class in sorted(df[target].unique()):
            group = df[df[target] == price_class][col].dropna()
            groups.append(group)

        f_stat, p_value = f_oneway(*groups)

        anova_results.append({
            "Feature": col,
            "F_statistic": f_stat,
            "p_value": p_value,
            "Significatif": p_value < 0.05
        })

anova_df = pd.DataFrame(anova_results).sort_values(by="p_value")

print(anova_df)

print("\nInterprétation ANOVA :")
print("Si p_value < 0.05, la variable est statistiquement différente selon les gammes de prix.")



correlation_results = []

for col in numeric_columns:
    if col != target:
        pearson_corr, pearson_p = pearsonr(df[col], df[target])
        spearman_corr, spearman_p = spearmanr(df[col], df[target])

        correlation_results.append({
            "Feature": col,
            "Pearson_corr": pearson_corr,
            "Pearson_p_value": pearson_p,
            "Spearman_corr": spearman_corr,
            "Spearman_p_value": spearman_p
        })

correlation_df = pd.DataFrame(correlation_results)
correlation_df = correlation_df.sort_values(by="Pearson_corr", ascending=False)

print("\nCorrélation des caractéristiques avec price_range :")
print(correlation_df)


plt.figure(figsize=(16, 12))

correlation_matrix = df.corr()

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=False
)

plt.title("Carte thermique des corrélations entre les variables")
plt.show()

target_corr = df.corr()[target].sort_values(ascending=False)

plt.figure(figsize=(10, 8))
target_corr.drop(target).plot(kind="barh")

plt.title("Corrélation des caractéristiques avec la gamme de prix")
plt.xlabel("Coefficient de corrélation")
plt.ylabel("Caractéristiques")
plt.grid(True)
plt.show()

print("\nVariables les plus corrélées avec price_range :")
print(target_corr.head(10))


important_features = target_corr.drop(target).abs().sort_values(ascending=False).head(6).index

print("\nVariables importantes sélectionnées pour visualisation :")
print(important_features)

for col in important_features:
    plt.figure(figsize=(8, 5))
    plt.hist(df[col], bins=30)
    plt.title(f"Distribution de {col}")
    plt.xlabel(col)
    plt.ylabel("Fréquence")
    plt.grid(True)
    plt.show()

for col in important_features:
    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x=target,
        y=col
    )

    plt.title(f"{col} selon la gamme de prix")
    plt.xlabel("Gamme de prix")
    plt.ylabel(col)
    plt.show()


if "ram" in df.columns and "battery_power" in df.columns:
    plt.figure(figsize=(8, 6))

    sns.scatterplot(
        data=df,
        x="ram",
        y="battery_power",
        hue=target,
        palette="viridis"
    )

    plt.title("Relation entre RAM, Battery Power et Price Range")
    plt.xlabel("RAM")
    plt.ylabel("Battery Power")
    plt.show()

if "ram" in df.columns:
    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df,
        x="ram",
        y=target,
        hue=target,
        palette="viridis"
    )

    plt.title("Relation entre RAM et gamme de prix")
    plt.xlabel("RAM")
    plt.ylabel("Price Range")
    plt.show()


group_means = df.groupby(target).mean()

print("\nMoyennes des caractéristiques par gamme de prix :")
print(group_means)

plt.figure(figsize=(12, 8))

selected_cols = important_features.tolist()

group_means[selected_cols].plot(kind="bar", figsize=(12, 7))

plt.title("Moyennes des principales caractéristiques par gamme de prix")
plt.xlabel("Gamme de prix")
plt.ylabel("Valeur moyenne")
plt.xticks(rotation=0)
plt.grid(True)
plt.show()


plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x=target,
    palette="viridis"
)

plt.title("Répartition des gammes de prix")
plt.xlabel("Gamme de prix")
plt.ylabel("Nombre de téléphones")
plt.show()



if "ram" in df.columns:
    print("\nStatistiques de RAM par gamme de prix :")
    print(df.groupby(target)["ram"].describe())

    plt.figure(figsize=(8, 5))
    sns.boxplot(data=df, x=target, y="ram", palette="viridis")

    plt.title("RAM selon la gamme de prix")
    plt.xlabel("Gamme de prix")
    plt.ylabel("RAM")
    plt.show()


if "battery_power" in df.columns:
    print("\nStatistiques de battery_power par gamme de prix :")
    print(df.groupby(target)["battery_power"].describe())

    plt.figure(figsize=(8, 5))
    sns.boxplot(data=df, x=target, y="battery_power", palette="viridis")

    plt.title("Battery Power selon la gamme de prix")
    plt.xlabel("Gamme de prix")
    plt.ylabel("Battery Power")
    plt.show()


top_features = target_corr.drop(target).abs().sort_values(ascending=False).head(10)

print("\nTop 10 des facteurs les plus liés à la gamme de prix :")
print(top_features)

plt.figure(figsize=(10, 6))

top_features.sort_values().plot(kind="barh")

plt.title("Top 10 des facteurs déterminants de la gamme de prix")
plt.xlabel("Corrélation absolue avec price_range")
plt.ylabel("Caractéristiques")
plt.grid(True)
plt.show()


print("\n SYNTHÈSE ET CONCLUSION ")

print("\n1. Chargement et exploration")
print("Le dataset a été chargé avec Pandas. Sa structure, les types de données et les valeurs manquantes ont été examinés.")

print("\n2. Nettoyage et prétraitement")
print("Les doublons ont été supprimés. Les valeurs manquantes numériques ont été remplacées par la médiane et les valeurs catégorielles par le mode.")
print("Les variables catégorielles ont été transformées en variables numériques si elles existaient.")

print("\n3. Analyse statistique")
print("NumPy et SciPy ont été utilisés pour calculer la moyenne, la médiane, le mode, l'étendue, la variance, l'écart type, l'asymétrie et le kurtosis.")
print("Des tests ANOVA ont été réalisés pour vérifier si les caractéristiques diffèrent significativement selon les gammes de prix.")

print("\n4. Corrélations")
print("Les corrélations Pearson et Spearman ont été calculées entre chaque caractéristique et la variable cible price_range.")

print("\n5. Principaux facteurs déterminants")
print("Les variables les plus liées à la gamme de prix sont :")
print(top_features)

print("\n6. Conclusion générale")
print("Les caractéristiques ayant les corrélations les plus fortes avec price_range sont les meilleurs indicateurs pour expliquer la classification du prix des téléphones.")
print("Dans ce type de dataset, la RAM est souvent l'un des facteurs les plus déterminants de la gamme de prix.")
print("Les visualisations confirment les différences entre les groupes de prix et permettent d’identifier les variables les plus importantes.")
```
