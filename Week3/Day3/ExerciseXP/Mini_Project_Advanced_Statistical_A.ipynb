```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy import stats
from scipy.stats import ttest_ind, normaltest

plt.style.use("default")

file_path = "/content/Apple Stock Prices (1981 to 2023).csv"

df = pd.read_csv(file_path)

print("Aperçu du dataset :")
print(df.head())

print("\nDimensions du dataset :")
print(df.shape)

print("\nColonnes disponibles :")
print(df.columns)

print("\nInformations générales :")
print(df.info())

print("\nValeurs manquantes par colonne :")
print(df.isnull().sum())

df.columns = df.columns.str.strip()

print("\nColonnes après nettoyage :")
print(df.columns)

date_col = "Date"

possible_open = ["Open", "Ouverture"]
possible_high = ["High", "Haut"]
possible_low = ["Low", "Bas"]
possible_close = ["Close", "Clôture", "Close/Last"]
possible_adj_close = ["Adj Close", "Adjusted Close", "Cours de clôture ajusté"]
possible_volume = ["Volume"]

def find_column(possible_names):
    for name in possible_names:
        if name in df.columns:
            return name
    return None

open_col = find_column(possible_open)
high_col = find_column(possible_high)
low_col = find_column(possible_low)
close_col = find_column(possible_close)
adj_close_col = find_column(possible_adj_close)
volume_col = find_column(possible_volume)

print("\nColonnes détectées :")
print("Date :", date_col)
print("Open :", open_col)
print("High :", high_col)
print("Low :", low_col)
print("Close :", close_col)
print("Adj Close :", adj_close_col)
print("Volume :", volume_col)


df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

numeric_cols = [open_col, high_col, low_col, close_col, adj_close_col, volume_col]
numeric_cols = [col for col in numeric_cols if col is not None]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=[date_col, close_col])

df = df.sort_values(by=date_col)

df["Year"] = df[date_col].dt.year
df["Month"] = df[date_col].dt.month
df["Year_Month"] = df[date_col].dt.to_period("M")

print("\nAperçu après nettoyage :")
print(df.head())

print("\nTypes après conversion :")
print(df.dtypes)

print("\nPériode couverte par les données :")
print("Date de début :", df[date_col].min())
print("Date de fin :", df[date_col].max())

print("\nNombre d'années disponibles :")
print(df["Year"].nunique())

print("\nFréquence approximative des données :")
print(df[date_col].diff().value_counts().head())


plt.figure(figsize=(14, 6))
plt.plot(df[date_col], df[close_col], linewidth=1)

plt.title("Évolution du cours de clôture d'Apple AAPL")
plt.xlabel("Date")
plt.ylabel("Cours de clôture")
plt.grid(True)
plt.show()



if volume_col is not None:
    plt.figure(figsize=(14, 6))
    plt.plot(df[date_col], df[volume_col], linewidth=1)

    plt.title("Évolution du volume échangé d'Apple AAPL")
    plt.xlabel("Date")
    plt.ylabel("Volume")
    plt.grid(True)
    plt.show()



if open_col is not None and high_col is not None and low_col is not None and close_col is not None:
    candle_data = df.tail(60).copy()
    candle_data = candle_data.reset_index(drop=True)

    plt.figure(figsize=(15, 6))

    for i in range(len(candle_data)):
        open_price = candle_data.loc[i, open_col]
        close_price = candle_data.loc[i, close_col]
        high_price = candle_data.loc[i, high_col]
        low_price = candle_data.loc[i, low_col]

        color = "green" if close_price >= open_price else "red"

        plt.plot([i, i], [low_price, high_price], color=color, linewidth=1)

        plt.plot([i, i], [open_price, close_price], color=color, linewidth=5)

    plt.title("Graphique en chandeliers simplifié - 60 dernières séances")
    plt.xlabel("Jours récents")
    plt.ylabel("Prix")
    plt.grid(True)
    plt.show()

key_columns = [open_col, high_col, low_col, close_col, adj_close_col, volume_col]
key_columns = [col for col in key_columns if col is not None]

print("\nStatistiques descriptives des colonnes clés :")
print(df[key_columns].describe())

print("\nMoyenne des colonnes clés :")
print(df[key_columns].mean())

print("\nMédiane des colonnes clés :")
print(df[key_columns].median())

print("\nÉcart type des colonnes clés :")
print(df[key_columns].std())

df["MA_20"] = df[close_col].rolling(window=20).mean()
df["MA_50"] = df[close_col].rolling(window=50).mean()
df["MA_200"] = df[close_col].rolling(window=200).mean()

plt.figure(figsize=(14, 6))
plt.plot(df[date_col], df[close_col], label="Close", linewidth=1)
plt.plot(df[date_col], df["MA_20"], label="Moyenne mobile 20 jours", linewidth=1.5)
plt.plot(df[date_col], df["MA_50"], label="Moyenne mobile 50 jours", linewidth=1.5)
plt.plot(df[date_col], df["MA_200"], label="Moyenne mobile 200 jours", linewidth=1.5)

plt.title("Cours de clôture AAPL avec moyennes mobiles")
plt.xlabel("Date")
plt.ylabel("Prix")
plt.legend()
plt.grid(True)
plt.show()

df["Daily_Return"] = df[close_col].pct_change()

print("\nAperçu des rendements quotidiens :")
print(df[[date_col, close_col, "Daily_Return"]].head())

plt.figure(figsize=(12, 6))
plt.hist(df["Daily_Return"].dropna(), bins=80)

plt.title("Distribution des rendements quotidiens d'AAPL")
plt.xlabel("Rendement quotidien")
plt.ylabel("Fréquence")
plt.grid(True)
plt.show()


available_years = sorted(df["Year"].dropna().unique())

print("\nAnnées disponibles :")
print(available_years[:5], "...", available_years[-5:])

year_1 = available_years[-2]
year_2 = available_years[-1]

close_year_1 = df[df["Year"] == year_1][close_col].dropna()
close_year_2 = df[df["Year"] == year_2][close_col].dropna()

t_stat, p_value = ttest_ind(close_year_1, close_year_2, equal_var=False)

print(f"\nTest t entre les cours de clôture moyens de {year_1} et {year_2} :")
print("Moyenne", year_1, ":", close_year_1.mean())
print("Moyenne", year_2, ":", close_year_2.mean())
print("Statistique t :", t_stat)
print("Valeur p :", p_value)

if p_value < 0.05:
    print("Conclusion : Il existe une différence significative entre les cours moyens des deux années.")
else:
    print("Conclusion : Il n'existe pas de différence significative entre les cours moyens des deux années.")

returns = df["Daily_Return"].dropna()

normal_stat, normal_p = normaltest(returns)

print("\nTest de normalité des rendements quotidiens :")
print("Statistique :", normal_stat)
print("Valeur p :", normal_p)

if normal_p < 0.05:
    print("Conclusion : Les rendements quotidiens ne suivent probablement pas une distribution normale.")
else:
    print("Conclusion : Les rendements quotidiens peuvent être considérés comme proches d'une distribution normale.")


close_values = df[close_col].dropna().to_numpy()

window = 30
weights = np.ones(window) / window

moving_average_np = np.convolve(close_values, weights, mode="valid")

print("\nMoyenne mobile calculée avec NumPy convolve :")
print(moving_average_np[:10])

plt.figure(figsize=(14, 6))
plt.plot(df[date_col], df[close_col], label="Close", alpha=0.5)

dates_ma = df[date_col].iloc[window-1:]

plt.plot(dates_ma, moving_average_np, label="Moyenne mobile 30 jours avec np.convolve", linewidth=2)

plt.title("Moyenne mobile avec NumPy convolve")
plt.xlabel("Date")
plt.ylabel("Prix")
plt.legend()
plt.grid(True)
plt.show()

corr_columns = [close_col, volume_col, "MA_20", "MA_50", "MA_200", "Daily_Return"]
corr_columns = [col for col in corr_columns if col is not None and col in df.columns]

corr_data = df[corr_columns].dropna()

corr_matrix = np.corrcoef(corr_data.to_numpy(), rowvar=False)

print("\nMatrice de corrélation avec NumPy :")
print(corr_matrix)

plt.figure(figsize=(8, 6))
plt.imshow(corr_matrix, cmap="coolwarm")
plt.colorbar(label="Corrélation")

plt.xticks(range(len(corr_columns)), corr_columns, rotation=45)
plt.yticks(range(len(corr_columns)), corr_columns)

plt.title("Matrice de corrélation entre indicateurs financiers")

for i in range(len(corr_columns)):
    for j in range(len(corr_columns)):
        plt.text(j, i, f"{corr_matrix[i, j]:.2f}", ha="center", va="center")

plt.tight_layout()
plt.show()



if volume_col is not None:
    ma_volume_corr = df[["MA_20", "MA_50", "MA_200", volume_col]].dropna().corr()

    print("\nCorrélation entre moyennes mobiles et volume :")
    print(ma_volume_corr)

from scipy.signal import savgol_filter

smooth_close = savgol_filter(df[close_col].dropna(), window_length=51, polyorder=3)

plt.figure(figsize=(14, 6))
plt.plot(df[date_col], df[close_col], label="Close original", alpha=0.4)
plt.plot(df[date_col], smooth_close, label="Close lissé avec SciPy", linewidth=2)

plt.title("Lissage du cours de clôture avec SciPy")
plt.xlabel("Date")
plt.ylabel("Prix")
plt.legend()
plt.grid(True)
plt.show()


print("\n================ SUMMARY AND INSIGHTS ================")

print("\n1. Data Loading and Exploration")
print("Le dataset AAPL a été chargé avec Pandas. Les dates ont été converties au format datetime et les colonnes numériques ont été nettoyées.")

print("\n2. Data Visualization")
print("Les graphiques montrent l'évolution du cours de clôture, du volume échangé et des prix sous forme de chandeliers simplifiés.")

print("\n3. Statistical Analysis")
print(f"Moyenne du cours de clôture : {df[close_col].mean():.2f}")
print(f"Médiane du cours de clôture : {df[close_col].median():.2f}")
print(f"Écart type du cours de clôture : {df[close_col].std():.2f}")

print("\n4. Hypothesis Testing")
print(f"Le test t a comparé les cours de clôture moyens entre {year_1} et {year_2}.")
print(f"Valeur p du test t : {p_value}")

print("\n5. Advanced Statistical Techniques")
print("NumPy a été utilisé avec convolve pour calculer une moyenne mobile.")
print("np.corrcoef a été utilisé pour analyser les corrélations entre le cours de clôture, le volume, les moyennes mobiles et les rendements.")
print("SciPy a été utilisé pour tester la normalité des rendements et lisser le signal du cours de clôture.")

print("\n6. Conclusion")
print("L'action Apple montre une forte tendance de croissance sur le long terme.")
print("Les moyennes mobiles permettent d'observer les tendances générales malgré les fluctuations quotidiennes.")
print("Les rendements quotidiens ne suivent généralement pas une distribution normale parfaite, ce qui est fréquent dans les données financières.")

print("\n7. Reflection")
print("Les principaux défis étaient la gestion des dates, la détection automatique des colonnes et la visualisation de longues séries temporelles.")
print("Ces problèmes ont été résolus en nettoyant les colonnes, en convertissant les dates et en utilisant des moyennes mobiles pour simplifier l'analyse.")
```
