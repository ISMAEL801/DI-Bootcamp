```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

cities = [
    "City 1", "City 2", "City 3", "City 4", "City 5",
    "City 6", "City 7", "City 8", "City 9", "City 10"
]

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

temperature_data = np.random.uniform(low=-5, high=35, size=(10, 12))

temperature_data = np.round(temperature_data, 2)

df = pd.DataFrame(
    data=temperature_data,
    index=cities,
    columns=months
)

print("Dataset des températures mensuelles :")
print(df)


df["Annual Average"] = df.mean(axis=1)

print("\nTempérature moyenne annuelle pour chaque ville :")
print(df["Annual Average"])

hottest_city = df["Annual Average"].idxmax()
highest_average = df["Annual Average"].max()

coldest_city = df["Annual Average"].idxmin()
lowest_average = df["Annual Average"].min()

print("\nVille avec la température moyenne la plus élevée :")
print(hottest_city, "avec", round(highest_average, 2), "°C")

print("\nVille avec la température moyenne la plus basse :")
print(coldest_city, "avec", round(lowest_average, 2), "°C")


plt.figure(figsize=(14, 7))

for city in cities:
    plt.plot(
        months,
        df.loc[city, months],
        marker="o",
        linewidth=2,
        label=city
    )

plt.title("Monthly Temperature Trends for Each City", fontsize=18)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Temperature (°C)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(title="Cities", loc="center left", bbox_to_anchor=(1, 0.5))

plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))

df["Annual Average"].sort_values(ascending=False).plot(kind="bar")

plt.title("Température moyenne annuelle par ville")
plt.xlabel("Ville")
plt.ylabel("Température moyenne annuelle (°C)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()


print("\n================ RAPPORT FINAL ================")

print("\n1. Préparation des données")
print("Un jeu de données synthétiques a été généré avec NumPy.")
print("Les températures représentent les moyennes mensuelles de 10 villes sur 12 mois.")
print("Les valeurs sont comprises entre -5°C et 35°C.")

print("\n2. Analyse des températures")
print(f"La ville avec la température moyenne annuelle la plus élevée est {hottest_city}, avec {highest_average:.2f}°C.")
print(f"La ville avec la température moyenne annuelle la plus basse est {coldest_city}, avec {lowest_average:.2f}°C.")

print("\n3. Observation générale")
print("Le graphique linéaire permet de comparer les variations de température de chaque ville au fil des mois.")
print("Certaines villes présentent de fortes variations mensuelles, tandis que d'autres ont des températures plus stables.")
```
