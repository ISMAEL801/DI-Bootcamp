**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks

- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
quiz_prompt = '''Act as a 7th-grade science teacher. Your task is to create engaging educational content based on the provided article about volcanic eruptions. First, summarize the article in 2 simple bullet points, using vocabulary appropriate for 11-13 year olds. Second, generate 3 age-appropriate multiple-choice questions about the article. Each question must have one correct answer and two plausible, but incorrect, distractors. The overall tone should be friendly and encouraging. Finally, present all the content (summary and questions) in a clear, organized format suitable for direct integration into Google Slides, using headings for the summary and each question.'''
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
chosen_style = 'Functional'
prompt_template = '''As a customer support chatbot for an e-commerce company, analyze the following customer message and perform two tasks:

1.  **Summarize the customer's core issue** concisely in one sentence.
2.  **Propose a clear, step-by-step solution** or the next action the customer should take to resolve their issue. If the solution involves checking an order status, include a placeholder `[Order Number]`.

Customer Message: {customer_message}

Output Format:
Summary: [One-sentence summary of the issue]
Solution: [Step-by-step solution or next action]'''
key_style_features = [
    'Clearly defined tasks: The prompt explicitly asks for two specific actions (summarize and propose solution).',
    'Precise output format: It specifies the structure for the summary and solution, including placeholders, ensuring consistent, actionable responses.'
]
justification = '''The 'Functional' style is ideal for a customer support chatbot because its primary goal is to execute tasks with precision and provide direct, actionable responses. Unlike exploratory (brainstorming), conversational (mimicking human chat), or structured (fixed formatting for data entry) styles, a functional prompt ensures the chatbot efficiently processes input and delivers exact, solution-oriented output. This is crucial for resolving customer issues quickly and consistently, which directly impacts customer satisfaction in a support context.'''
```
