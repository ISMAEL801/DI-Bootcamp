## Step 1: Craft a Prompt for Tone, Format, and Length

First, let's set up the Gemini API. If you haven't already, please store your `GOOGLE_API_KEY` in Colab's secrets manager.

```python
import google.generativeai as genai
from google.colab import userdata

# Configure the Gemini API
GOOGLE_API_KEY=userdata.get('GOOGLE_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY)

# Initialize the Generative Model
gemini_model = genai.GenerativeModel('gemini-2.2-flash')
```

Now, let's define the input text and craft the prompt for rewriting it in a friendly, clear tone, using bullet points, paraphrasing, and staying under 75 words.

```python
input_text = "Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

prompt_step1 = f"""Rewrite the following text for an internal employee newsletter. The tone should be friendly and clear. Use bullet points to organize the information. Paraphrase the original content, avoiding direct quotations. The total output should be under 75 words.\n\nText: {input_text}"""

response_step1 = gemini_model.generate_content(prompt_step1)
print(response_step1.text)
```

## Step 2: Evaluate the Output

Let's review the generated output against the following criteria:

*   **Relevance**: Does the output stick to the message of the source content?
*   **Clarity**: Is the message easier to understand for non-technical staff?
*   **Structure**: Is the information formatted into clear bullet points?
*   **Tone**: Is the language appropriate for internal employee communication?
*   **Length**: Does it stay under 75 words?
*   **Factual Accuracy**: Did the model make up or remove any critical security details?

## Step 4: Paraphrasing Deep Dive

Now, let's create a second version of the prompt that paraphrases the policy paragraph for a junior intern audience. This version should use plain language and short phrases, with no more than 4 bullet points, and avoid any corporate or legal jargon. The tone should be supportive and informative.

```python
prompt_step4 = f"""Rewrite the following policy paragraph for a junior intern audience. Use plain language and short phrases. Use no more than 4 bullet points. Avoid any corporate or legal jargon. Keep the tone supportive and informative.\n\nText: {input_text}"""

response_step4 = gemini_model.generate_content(prompt_step4)
print(response_step4.text)
```

## Step 5: Quote Extraction Variant

Sometimes, quoting is more effective than paraphrasing. Let's extract one direct quote from the original text that best captures the core security policy.

```python
prompt_step5_quote = f"""From the following text, extract one direct quote that best captures the core security policy. Only provide the quote, nothing else.\n\nText: {input_text}"""

response_step5_quote = gemini_model.generate_content(prompt_step5_quote)
print(response_step5_quote.text)
```

### In what kind of internal communication would quoting be more appropriate than paraphrasing?

*   **Your Answer:** Quoting is more appropriate when the exact wording is critical, legally binding, or when referring to official policy statements, rules, or direct communications from leadership where precision and authenticity are paramount. It ensures no meaning is lost in interpretation.

### When might quoting pose a risk?

*   **Your Answer:** Quoting might pose a risk if the original text is complex, jargon-filled, or could be easily misinterpreted by the target audience. It can also be risky if the quote is taken out of context, if the language is not suitable for the audience, or if it makes the communication unnecessarily lengthy or formal.
